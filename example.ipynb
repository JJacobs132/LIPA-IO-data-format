```python
import numpy as np
import zipfile as zip
from datetime import datetime as dt
import matplotlib.pyplot as plt
from main import *
import os

def ricker(t,sig,deltat): #returns ricker pulse with width dependent on sig and offset by deltat
    return 2/(np.sqrt(3*sig)*np.pi**(1/4)) * (1-((t-deltat)/sig)**2) * np.exp(-1*(t-deltat)**2/(2*sig**2))
```

```python
# generate and write test DC
t = np.linspace(0,5,6)
DCa = np.meshgrid(t,t)[0] * np.meshgrid(t,t)[1]   

DCt = DC(DCa)
DCt.writenpy(file='test-dc')
```

```python
# generate and write test SPM
SPM_t = SPM(2,4)
for i in range(SPM_t.N_source):
    for j in range(SPM_t.N_reciever):
        current_time = str(dt.now())
        SPM_t.SPM[i,j]['t0'] = current_time
        SPM_t.SPM[i,j]['seqnr'] = i*SPM_t.N_reciever + j

# generate and write test CPSM
CSPM_t = CSPM(2,4)
CSPM_t.CSPM[0,1]['r_s'] = 1
CSPM_t.CSPM[0,2]['r_s'] = 10
CSPM_t.CSPM[0,2]['r_r'] = 100

SPM_t.writecsv('test-spm.csv')
CSPM_t.writecsv('test-cspm.csv')
```
